# Navigation
```bash
pwd                    # Show the current directory
ls                     # List directory contents
ls -l                  # Display a detailed listing
ls -a                  # Include hidden items
cd directoryname       # Enter a directory
cd ..                  # Move up one level (aka parent directory)
cd ../..               # Move up two levels
cd                     # Return to the home directory
cd ~                   # Return to the home directory
cd -                   # Return to the previous directory
```

# Finding files
```bash
find .                              # List everything (files and directories)
find . -name "07.txt"               # Find an exact name
find .. -name "07.txt"              # Find inside the parent directory an exact name
find ~ -name "07.txt"               # Find from your home directory an exact name
find . -type f                      # Find only for regular files
find . -type f -name "*.txt"        # Find text files
find . -type f -iname "*.txt"       # Ignore capitalization
find . -type f -name "0.txt"        # Find files whose names begin with 0
find . -type f -name "1*.txt"       # Find files containing the digit 1 anywhere in the name
```

## Wildcards
```bash
*    zero or more characters
?    exactly one character
```

## Examples
```bash
find . -name "*.txt"
find . -name "0?.txt"
find . -name "*1*.txt"
```
